```python
# CHANGE ACCORDINGLY: Define your mapping of CSV file names to game titles
UPDATED_METACRITIC_DIR="metacritic_reviews/updated_csv_files"

UPDATED_STEAM_DIR="steam_reviews/updated_csv_files"

UPDATED_REDDIT_DIR="reddit_reviews/updated_csv_files"

OUTPUT_DIR = "consolidated_reviews"

UPDATED_DIRS = {
                "metacritic": UPDATED_METACRITIC_DIR,
                # "steam": UPDATED_STEAM_DIR,
                # "reddit": UPDATED_REDDIT_DIR
                }

for platform, base_dir in UPDATED_DIRS.items():
  csv_files = glob.glob(f"{base_dir}/*.csv")

  # Create an empty list to store dataframes
  dataframes = []

  # Loop through each file and read it
  for file in csv_files:
    print(file)
    df = pd.read_csv(file, lineterminator='\n')
    if platform == "metacritic":
      df["date"] = pd.to_datetime(df["date"]).dt.strftime("%Y-%m-%d")
      df = df.rename(columns={'date': 'timestamp_updated_date','review': 'review_text'})
      df['score'] = df['score'].replace('tbd', pd.NA)
    elif platform == "steam":
        df = df[['author_steamid','review_text','votes_up', 'votes_funny', 'weighted_vote_score','platform', 'game', 'timestamp_updated_date']]
    elif platform == "reddit":
      df = df[['Title', 'URL', 'Score', 'review_text', 'Subreddit', 'Comment Author', 'Level', 'platform', 'game', 'timestamp_updated_date']]

    # Append the dataframe to the list
    dataframes.append(df)

  # Concatenate all dataframes, aligning columns automatically
  merged_df = pd.concat(dataframes, ignore_index=True, sort=False)

  print("All CSV files have been merged successfully!")

  # Save the updated dataframe to a new CSV file, just REWRITE
  output_file = os.path.join(OUTPUT_DIR, f"combined_{platform}.csv")
  merged_df.to_csv(output_file, index=False)
  print(f"Saved updated file: {output_file}")
```

```python
OUTPUT_DIR = "consolidated_reviews"
csv_file = "consolidated_reviews/combined_metacritic.csv"

df = pd.read_csv(csv_file)

# Define rating_class based on score
def classify_rating(score):
    if isinstance(score, str) and score.lower() == "tbd":
        return "tbd"
    elif pd.notna(score):
        score = int(score)
        if 0 <= score <= 39:
            return "negative"
        elif 40 <= score <= 69:
            return "neutral"
        elif 70 <= score <= 100:
            return "positive"
    return "unknown"  # For any unexpected values

df["rating_class"] = df["score"].apply(classify_rating)


# Save the updated dataframe to a new CSV file
output_file = os.path.join(OUTPUT_DIR, "combined_metacritic_score.csv")
df.to_csv(output_file, index=False)
print(f"Saved updated file: {output_file}")
```
