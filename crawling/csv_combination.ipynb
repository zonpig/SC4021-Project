```python
# CHANGE ACCORDINGLY: Define your mapping of CSV file names to game titles
UPDATED_METACRITIC_DIR="metacritic_reviews/updated_csv_files"

UPDATED_STEAM_DIR="steam_reviews/updated_csv_files"

UPDATED_REDDIT_DIR="reddit_reviews/updated_csv_files"

OUTPUT_DIR = "consolidated_reviews"

UPDATED_DIRS = {
                # "metacritic": UPDATED_METACRITIC_DIR,
                "steam": UPDATED_STEAM_DIR,
                # "reddit": UPDATED_REDDIT_DIR
                }

for platform, base_dir in UPDATED_DIRS.items():
  csv_files = glob.glob(f"{base_dir}/*.csv")

  # Create an empty list to store dataframes
  dataframes = []

  # Loop through each file and read it
  for file in csv_files:
    df = pd.read_csv(file, lineterminator='\n')
    # Append the dataframe to the list
    dataframes.append(df)

  # Concatenate all dataframes, aligning columns automatically
  merged_df = pd.concat(dataframes, ignore_index=True, sort=False)

  print("All CSV files have been merged successfully!")

  # Save the updated dataframe to a new CSV file, just REWRITE
  output_file = os.path.join(OUTPUT_DIR, f"combined_{platform}.csv")
  merged_df.to_csv(output_file, index=False)
  print(f"Saved updated file: {output_file}")
```

### Consolidating all platforms together

```python
OUTPUT_DIR = "./consolidated_reviews"
csv_files = glob.glob(f"{OUTPUT_DIR}/*.csv")
dataframes = []

# Iterate over each CSV file
for file in csv_files:
    df = pd.read_csv(file, lineterminator='\n')
    # Append the dataframe to the list
    dataframes.append(df)
    
merged_df = pd.concat(dataframes, ignore_index=True)

print("All CSV files have been merged successfully!")

# Save the updated dataframe to a new CSV file, just REWRITE
output_file = "combined_data.csv"
merged_df.to_csv(output_file, index=False)
print(f"Saved updated file: {output_file}")
```

```python
# Load CSV with correct line terminator
combined_data = pd.read_csv("combined_data.csv", lineterminator='\n')

# Total number of words in all 'review_text' cells
def count_total_words(df, column_name):
    if column_name in df.columns:
        # Drop NaNs, split text into words, flatten, and count total words
        total_words = df[column_name].dropna().str.split().explode().shape[0]
        return total_words
    else:
        print(f"Column '{column_name}' does not exist in the DataFrame.")
        return None

# Number of unique words in all 'review_text' cells
def count_unique_words(df, column_name):
    if column_name in df.columns:
        # Drop NaNs, split text into words, flatten, and get unique word count
        unique_words = df[column_name].dropna().str.split().explode().nunique()
        return unique_words
    else:
        print(f"Column '{column_name}' does not exist in the DataFrame.")
        return None

total_words = count_total_words(combined_data, 'review_text')
unique_words = count_unique_words(combined_data, 'review_text')

print(f"Total words: {total_words}")
print(f"Unique words: {unique_words}")
```
